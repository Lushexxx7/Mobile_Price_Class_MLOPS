# Entrenamiento y selecciÃ³n de modelos

ComparaciÃ³n reproducible de RegresiÃ³n LogÃ­stica, Random Forest y SVM mediante clases con una interfaz comÃºn.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.config import MODEL_PATH, TRAIN_PATH
from src.data.load_data import CargadorDatos
from src.models.train import ModeloRandomForest
from src.models.pipeline import PipelineTelefonos
```

## 1. Carga de datos

```python
datos = CargadorDatos(TRAIN_PATH).cargar()
print(f'Dimensiones: {datos.shape}')
datos.head()
```

## 2. Entrenamiento polimÃ³rfico

```python
pipeline = PipelineTelefonos(metrica_seleccion='accuracy')
resultados = pipeline.entrenar(datos)
resultados
```

```python
resultados.set_index('modelo')[['accuracy', 'precision', 'recall', 'f1']].plot(
    kind='bar', figsize=(10, 5), ylim=(0, 1), rot=0
)
plt.title('ComparaciÃ³n de modelos')
plt.ylabel('PuntuaciÃ³n')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 3. Importancia de caracterÃ­sticas de Random Forest

```python
random_forest = next(
    modelo for modelo in pipeline.modelos
    if isinstance(modelo, ModeloRandomForest)
)
importancias = (
    pd.Series(random_forest.obtener_importancias(), index=pipeline.columnas)
    .sort_values(ascending=True)
)
importancias.plot(kind='barh', figsize=(9, 7), title='Importancia de caracterÃ­sticas')
plt.xlabel('Importancia')
plt.tight_layout()
plt.show()
importancias.sort_values(ascending=False).to_frame('importancia')
```

## 4. Persistencia del mejor modelo

```python
ruta_modelo = pipeline.guardar(MODEL_PATH)
print(f'Mejor modelo: {pipeline.mejor_modelo.nombre}')
print(f'Guardado en: {ruta_modelo}')
```
