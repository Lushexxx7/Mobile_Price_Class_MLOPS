# PredicciÃ³n con el modelo final

Este notebook reutiliza el artefacto entrenado; no vuelve a ajustar ningÃºn modelo.

```python
import sys
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.config import MODEL_PATH, TEST_PATH
from src.data.load_data import CargadorDatos
from src.models.pipeline import PipelineTelefonos
```

## 1. Carga del modelo y los datos

```python
artefacto = PipelineTelefonos.cargar(MODEL_PATH)
datos_test = CargadorDatos(TEST_PATH).cargar()
print(f'Modelo cargado: {artefacto["nombre"]}')
print(f'Datos de prueba: {datos_test.shape}')
```

## 2. Predicciones

```python
predicciones = PipelineTelefonos.predecir(datos_test, MODEL_PATH)
resultado = pd.DataFrame({
    'id': datos_test['id'] if 'id' in datos_test else range(1, len(datos_test) + 1),
    'price_range': predicciones,
})
resultado.head()
```

```python
assert len(resultado) == len(datos_test)
assert set(resultado['price_range']).issubset({0, 1, 2, 3})
resultado['price_range'].value_counts().sort_index()
```

## 3. Guardado de resultados

```python
ruta_salida = PROJECT_ROOT / 'data' / 'processed' / 'predicciones.csv'
ruta_salida.parent.mkdir(parents=True, exist_ok=True)
resultado.to_csv(ruta_salida, index=False)
print(f'Predicciones guardadas en: {ruta_salida}')
```
